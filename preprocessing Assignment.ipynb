```python
import string
import re
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

def text_preprocessing(text):
    if not isinstance(text, str):
        return ""
    text = text.lower()
    text = re.sub(r'\s+', ' ', text)
    text = re.sub(r'\d+', '', text)
    text = "".join([a for a in text if a not in string.punctuation])
    text = " ".join([a for a in text.split() if a not in stop_words])
    return text

data['cleaned_reviews'] = data['reviews'].apply(text_preprocessing)

data[['reviews', 'cleaned_reviews']].head()
```
